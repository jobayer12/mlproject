```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
df = pd.read_csv('economic_index.csv')
df.head()
```

```python
df.drop(columns=['Unnamed: 0', 'year', 'month'], inplace=True)
df.head()
```

```python
sns.pairplot(df)
```

```python
df.corr()
```

```python
# Independet and dependent features
X = df.iloc[:, :-1]
y = df.iloc[:, -1] 
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
```

```python
sns.regplot(data=df, x='interest_rate', y='index_price')
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
X_train = scaler.fit_transform(X_train)
X_test = scaler.fit_transform(X_test)
```

```python
from sklearn.linear_model import LinearRegression
regression = LinearRegression()
```

```python
regression.fit(X_train, y_train)
```

```python
# Cross validation
from sklearn.model_selection import cross_val_score
validation_score = cross_val_score(regression, X_train, y_train, cv=3, scoring='neg_mean_squared_error')
```

```python
np.mean(validation_score)
```

```python
y_pred = regression.predict(X_test)
```

```python
y_pred
```

```python
from sklearn.metrics import mean_squared_error,mean_absolute_error
```

```python
mse=mean_squared_error(y_test,y_pred)
mae=mean_absolute_error(y_test,y_pred)
rmse=np.sqrt(mse)
print(mse)
print(mae)
print(rmse)
```

```python
from sklearn.metrics import r2_score
score=r2_score(y_test,y_pred)
print(score)
```
